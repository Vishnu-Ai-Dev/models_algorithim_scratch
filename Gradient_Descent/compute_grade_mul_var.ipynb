```python
import numpy as np


X_train = np.array([
    [1000, 2, 10],
    [1500, 3,  5],
    [2000, 4,  2]
])

y_train = np.array([200, 300, 400])
```

```python
def compute_gradient(X, y, w, b):

    m, n = X.shape

    dj_dw = np.zeros(n)
    dj_db = 0

    for i in range(m):

        f_wb = np.dot(X[i], w) + b

        error = f_wb - y[i]

        for j in range(n):

            dj_dw[j] = dj_dw[j] + error * X[i, j]

        dj_db = dj_db + error

    dj_dw = dj_dw / m
    dj_db = dj_db / m

    return dj_dw, dj_db
```

```python
def gradient_descent(X, y, w, b, alpha, num_iters):

    for i in range(num_iters):

        dw, db = compute_gradient(X, y, w, b)

        w = w - alpha * dw
        b = b - alpha * db

    return w, b
```

```python
w = np.zeros(3)
b = 0

alpha = 0.000001
num_iters = 1000
```

```python
w, b = gradient_descent(
    X_train,
    y_train,
    w,
    b,
    alpha,
    num_iters
)
print(w)
print(b)
```
