```python
import numpy as np

def compute_cost(X, y, w, b):
    # 1. Get the number of training examples (m) by checking the rows of X
    m = X.shape[0] 
    
    # 2. Calculate the predictions for all rows at once using vectorization
    predictions = np.dot(X, w) + b 
    
    # 3. Calculate the squared errors
    squared_errors = (predictions - y) ** 2 
    
    # 4. Sum the errors and divide by 2m to get the final Cost (J)
    cost = np.sum(squared_errors) / (2 * m)
    
    return cost
```

```python
import numpy as np

def compute_gradient(X, y, w, b):
    # 1. Get the number of training examples
    m = X.shape[0]
    
    # 2. Get predictions (just like we did in the cost function)
    predictions = np.dot(X, w) + b
    
    # 3. Calculate the baseline error
    error = predictions - y
    
    # 4. Calculate the gradient for the weight (dj_dw)
    # Using the dot product multiplies the errors by X and sums them up automatically!
    dj_dw = np.dot(X.T, error) / m
    
    # 5. Calculate the gradient for the bias (dj_db)
    dj_db = np.sum(error) / m
    
    return dj_dw, dj_db
```
