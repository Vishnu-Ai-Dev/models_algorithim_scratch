Z score normalization

This is usally used when the scale of the features are so high so we need better scaling in order to make the line fit the graph better

x1≤x≤xn

```python
import numpy as np

x_train=np.array([500,600,880,1000,1200,1500,1800,2000])

print(x_train.shape[0])

print(x_train[7])

mean=np.mean(x_train,axis=0)
st=np.std(x_train,axis=0)

x_norm=(x_train-mean)/st

print(f"The normalized version of the training data is: {x_norm}")
```

```python
import numpy as np

x_train=np.array([500,600,880,1000,1200,1500,1800,2000])
last_sum=x_train[7]
init_sum=x_train[0]
mean=np.mean(x_train,axis=0)

nume=x_train[0]-mean
deno=init_sum-last_sum

mean_norm=(nume/deno)

print(f"The normalized version of the training data is: {mean_norm}")
```
