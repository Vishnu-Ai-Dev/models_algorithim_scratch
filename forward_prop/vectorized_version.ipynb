```python
import numpy as np
X=np.array([200,17])
w=np.array([[1,-3,5],[7,3,2]])
b=np.array([1,2,3])
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
def dense(a_in,w,b):
    z=np.matmul(a_in,w)+b
    aout=sigmoid(z)
    return aout
```

```python
output=dense(X,w,b)
print(output)
```
