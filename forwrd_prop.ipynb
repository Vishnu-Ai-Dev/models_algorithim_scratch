```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
def dense(a_in, W, b):
    units = W.shape[1]             
    a_out = np.zeros(units)
    for j in range(units):          
        w = W[:, j]             
        z = np.dot(w, a_in) + b[j]  
        a_out[j] = sigmoid(z)       
    print(a_out)
```
