```python
import numpy as np
```

```python
X=np.array([[[4.0, 80.0]]])
w_1=np.array([[[0.4,0.7],[0.7,0.8]]])
b_1=np.array([[[0.1,0.2]]])
w_2=np.array([[[0.3],[0.5]]])
b_2=np.array([[[0.3]]])
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
a_1=X@w_1+b_1
print("weighted sums:",a_1)
z_1=sigmoid(a_1)
a_2=z_1@w_2+b_2
print("weighted sums:",a_2)
z_2=sigmoid(a_2)
print("predictions:",z_2)

```
