```python
import numpy as np

X_train = np.array([
    [1, 50],
    [2, 55],
    [3, 60],
    [4, 70],
    [5, 75],
    [6, 80]
])

y_train = np.array([0, 0, 0, 1, 1, 1])

w = np.array([1.0, 0.1])
b = -8.0
```

```python
def sigmoid_function(x,y,w,b,z):
    m=x.shape
    z=np.dot(x,w)+b
    y_pred=1/(1+np.exp(-z))
    return y_pred
```

```python
predictions = sigmoid_function(X_train, y_train, w, b, 0)
print(predictions)
```
